```python
import os
from dotenv import load_dotenv
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-120b",groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import HumanMessage

model.invoke([HumanMessage(content="Hi, my name is tarun")])
```

```python
from langchain_core.messages import AIMessage

model.invoke(

    [

        HumanMessage(content="Hi, my name is tarun and i am learning ai"),
        AIMessage(content="Hello, Tarun! Nice to meet you. How can I help you today?"),
        HumanMessage(content="hey whats my name and what do i do?")
    ]
)
```

```python
### messaging history
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}
def get_session_history(session_id:str) ->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model,get_session_history)
```

```python
config = {"configurable":{"session_id":"chat1"}}
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="hi my name is tarun and i am learning ai")],
    config=config
)
```

```python
response.content
```

```python
#### change the config = session id
config3 = {"configurable":{"session_id":"chat3"}}
response = with_message_history.invoke(
    [HumanMessage(content="what is my name")],
    config=config3
)

response.content
```

```python
#### prompt

from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpfull assistant, answer all the question to the best of your abiltiy"),
        MessagesPlaceholder(variable_name="messages")
    ]
)
```

```python
chain = prompt | model
```

```python
chain.invoke({"messages":[HumanMessage(content="hi my name is tarun")]})
```

```python
with_message_history=RunnableWithMessageHistory(chain,get_session_history)
```

```python
config = {"configurable":{"session_id":"chat4"}}
```

```python
response=with_message_history.invoke(
    [HumanMessage(content="what is my name")],
    config=config
)
```

```python
response
```

```python
### add more cmplx

from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpfull assistant, answer all the question to the best of your abiltiy in this {language}"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt | model
```

```python
response= chain.invoke({"messages":[HumanMessage(content="hi my name is tarun")],"language":"kannada"})
response.content
```

```python
##### wrap more complcated chain

with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
config = {"configurable":{"session_id":"chat5"}}
response = with_message_history.invoke(
    {"messages":[HumanMessage(content="hi my name is tarun")],"language":"engish"},
    config=config
)
```

```python
response.content
```

```python
### managining conversation history


from langchain_core.messages import SystemMessage,trim_messages

trimmer = trim_messages(
    max_tokens=70,
    strategy="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
```

```python
messages = [
    HumanMessage(content="Hi"),
    AIMessage(content="Hello"),
    HumanMessage(content="I like vanilla ice cream")
]
```

```python
from operator import itemgetter

from langchain_core.runnables import RunnablePassthrough

chain = (
    RunnablePassthrough.assign(messages=itemgetter("messages")|trimmer) | prompt | model
)


chain.invoke(
    {
    "messages":messages+ [HumanMessage(content="what ice creamm do i like")],
    "language":"english"

    }

)

response.content
```
